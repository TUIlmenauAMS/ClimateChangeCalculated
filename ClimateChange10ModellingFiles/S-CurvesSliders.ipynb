```python
import numpy as np
import matplotlib.pyplot as plt

from ipywidgets import interact, FloatSlider, IntSlider

def logistic(t, L, k, t0):
    return L / (1 + np.exp(-k * (t - t0)))

t_min, t_max = 2000, 2040
t = np.linspace(t_min, t_max, 800)

@interact(
    t0=FloatSlider(value=2025, min=t_min, max=t_max, step=0.1, description="t0"),
    k=FloatSlider(value=0.4, min=0.01, max=2.0, step=0.01, description="k"),
    L=FloatSlider(value=100, min=10, max=110, step=1, description="L (%)"),
    show_points=IntSlider(value=0, min=0, max=1, step=1, description="show t0"),
)
def plot_scurve(t0, k, L, show_points):
    y = logistic(t, L, k, t0)

    plt.figure(figsize=(10, 5))
    plt.plot(t, y, linewidth=3)
    plt.ylim(0, max(110, L))
    plt.xlim(t_min, t_max)
    plt.xlabel("Year (t)")
    plt.ylabel("Adoption (%)")
    plt.title("Interactive S-curve: vary L, k, and t0")
    plt.grid(True, linestyle=":", alpha=0.6)

    if show_points:
        plt.axvline(t0, linestyle="--", linewidth=2)
        plt.scatter([t0], [L/2], s=80)
        plt.text(t0 + 0.2, L/2, "  inflection (50%)", va="center")

    plt.show()
```
